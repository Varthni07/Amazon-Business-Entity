```python
# Load test datasets

test_s1 = pd.read_csv(f"{BASE}/test_source1.tsv", sep="\t")
test_s2 = pd.read_csv(f"{BASE}/test_source2.tsv", sep="\t")
test_s3 = pd.read_csv(f"{BASE}/test_source3.tsv", sep="\t")

print("Test Source 1:", test_s1.shape)
print("Test Source 2:", test_s2.shape)
print("Test Source 3:", test_s3.shape)
```

```python
# Basic text normalization

def normalize_basic(series):
    return (
        series.fillna("")
        .astype(str)
        .str.lower()
        .str.replace("&", " and ", regex=False)
        .str.replace(r"[^\w\s]", " ", regex=True)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

# Normalize names
train_s1["name_norm"] = normalize_basic(train_s1["business_name"])
train_s2["name_norm"] = normalize_basic(train_s2["business_name"])
train_s3["name_norm"] = normalize_basic(train_s3["business_name"])

# Normalize addresses
train_s1["address_norm"] = normalize_basic(train_s1["business_address"])
train_s2["address_norm"] = normalize_basic(train_s2["business_address"])
train_s3["address_norm"] = normalize_basic(train_s3["business_address"])

# Normalize countries
train_s1["country_norm"] = normalize_basic(train_s1["country"])
train_s2["country_norm"] = normalize_basic(train_s2["country"])
train_s3["country_norm"] = normalize_basic(train_s3["country"])

# Do the same for test data
test_s1["name_norm"] = normalize_basic(test_s1["business_name"])
test_s2["name_norm"] = normalize_basic(test_s2["business_name"])
test_s3["name_norm"] = normalize_basic(test_s3["business_name"])

test_s1["address_norm"] = normalize_basic(test_s1["business_address"])
test_s2["address_norm"] = normalize_basic(test_s2["business_address"])
test_s3["address_norm"] = normalize_basic(test_s3["business_address"])

test_s1["country_norm"] = normalize_basic(test_s1["country"])
test_s2["country_norm"] = normalize_basic(test_s2["country"])
test_s3["country_norm"] = normalize_basic(test_s3["country"])

print("Normalization completed successfully.")
```

```python
# Remove common legal/business suffixes from names

import re

def normalize_name(series):
    s = normalize_basic(series)

    # Remove common business suffixes at the end
    s = s.str.replace(
        r"\b(incorporated|inc|llc|ltd|limited|corp|corporation|company|co|pvt|plc)\b$",
        "",
        regex=True
    )

    return s.str.strip()

# Apply to training data
train_s1["name_norm"] = normalize_name(train_s1["business_name"])
train_s2["name_norm"] = normalize_name(train_s2["business_name"])
train_s3["name_norm"] = normalize_name(train_s3["business_name"])

# Apply to test data
test_s1["name_norm"] = normalize_name(test_s1["business_name"])
test_s2["name_norm"] = normalize_name(test_s2["business_name"])
test_s3["name_norm"] = normalize_name(test_s3["business_name"])

print("Business name normalization completed successfully.")
```

```python
# Create blocking keys

# Source 1
train_s1["name_key"] = train_s1["country_norm"] + "|" + train_s1["name_norm"]
train_s1["name_prefix_key"] = (
    train_s1["country_norm"] + "|" + train_s1["name_norm"].str[:4]
)
train_s1["address_key"] = train_s1["country_norm"] + "|" + train_s1["address_norm"]

# Source 2
train_s2["name_key"] = train_s2["country_norm"] + "|" + train_s2["name_norm"]
train_s2["name_prefix_key"] = (
    train_s2["country_norm"] + "|" + train_s2["name_norm"].str[:4]
)
train_s2["address_key"] = train_s2["country_norm"] + "|" + train_s2["address_norm"]

# Source 3
train_s3["name_key"] = train_s3["country_norm"] + "|" + train_s3["name_norm"]
train_s3["name_prefix_key"] = (
    train_s3["country_norm"] + "|" + train_s3["name_norm"].str[:4]
)
train_s3["address_key"] = train_s3["country_norm"] + "|" + train_s3["address_norm"]

# Test Source 1
test_s1["name_key"] = test_s1["country_norm"] + "|" + test_s1["name_norm"]
test_s1["name_prefix_key"] = (
    test_s1["country_norm"] + "|" + test_s1["name_norm"].str[:4]
)
test_s1["address_key"] = test_s1["country_norm"] + "|" + test_s1["address_norm"]

# Test Source 2
test_s2["name_key"] = test_s2["country_norm"] + "|" + test_s2["name_norm"]
test_s2["name_prefix_key"] = (
    test_s2["country_norm"] + "|" + test_s2["name_norm"].str[:4]
)
test_s2["address_key"] = test_s2["country_norm"] + "|" + test_s2["address_norm"]

# Test Source 3
test_s3["name_key"] = test_s3["country_norm"] + "|" + test_s3["name_norm"]
test_s3["name_prefix_key"] = (
    test_s3["country_norm"] + "|" + test_s3["name_norm"].str[:4]
)
test_s3["address_key"] = test_s3["country_norm"] + "|" + test_s3["address_norm"]

print("Blocking keys created successfully.")
```

```python
# Create candidate pairs using blocking

def make_candidates(left, right):
    candidates = []

    for key_col in ["name_key", "name_prefix_key", "address_key"]:
        # Count how many records share each key on the right side
        right_counts = right[key_col].value_counts()

        # Keep only useful, small blocks
        valid_keys = right_counts[
            (right_counts <= 5) & (right_counts.index != "")
        ].index

        # Select matching blocks
        left_block = left[left[key_col].isin(valid_keys)]
        right_block = right[right[key_col].isin(valid_keys)]

        # Merge records that share the blocking key
        pairs = left_block[
            ["entity_id", key_col]
        ].merge(
            right_block[
                ["entity_id", key_col]
            ],
            on=key_col,
            suffixes=("_left", "_right")
        )

        candidates.append(
            pairs[["entity_id_left", "entity_id_right"]]
        )

    # Combine candidates from all blocking methods
    result = pd.concat(candidates, ignore_index=True)

    # Remove duplicate candidate pairs
    result = result.drop_duplicates()

    return result


print("Generating training candidates...")

train_candidates_s1_s2 = make_candidates(train_s1, train_s2)
train_candidates_s1_s3 = make_candidates(train_s1, train_s3)

print("S1-S2 candidates:", len(train_candidates_s1_s2))
print("S1-S3 candidates:", len(train_candidates_s1_s3))
```

```python
# Combine S1-S2 and S1-S3 candidate pairs

train_candidates = pd.concat(
    [
        train_candidates_s1_s2.assign(source="S2"),
        train_candidates_s1_s3.assign(source="S3")
    ],
    ignore_index=True
)

print("Total candidate pairs:", len(train_candidates))
print("S2 candidates:", len(train_candidates_s1_s2))
print("S3 candidates:", len(train_candidates_s1_s3))
```

```python
# Create positive match pairs from ground truth

# Split the comma-separated matched IDs
ground_truth_pairs = ground_truth.copy()

ground_truth_pairs["matched_entity_ids"] = (
    ground_truth_pairs["matched_entity_ids"]
    .fillna("")
    .astype(str)
    .str.split(",")
)

# One row per actual match
ground_truth_pairs = ground_truth_pairs.explode("matched_entity_ids")

# Remove empty matches
ground_truth_pairs = ground_truth_pairs[
    ground_truth_pairs["matched_entity_ids"].str.strip() != ""
]

# Create a lookup set for fast checking
true_pairs = set(
    zip(
        ground_truth_pairs["source1_entity_id"],
        ground_truth_pairs["matched_entity_ids"]
    )
)

print("Ground-truth positive pairs:", len(true_pairs))

# Label each candidate
train_candidates["label"] = [
    1 if (s1, s2) in true_pairs else 0
    for s1, s2 in zip(
        train_candidates["entity_id_left"],
        train_candidates["entity_id_right"]
    )
]

print("Positive candidate pairs:", train_candidates["label"].sum())
print("Negative candidate pairs:", (train_candidates["label"] == 0).sum())
```

```python
# Create matching features

def add_features(candidates, left, right):
    left_data = left[
        [
            "entity_id",
            "name_norm",
            "address_norm",
            "country_norm"
        ]
    ].drop_duplicates("entity_id")

    right_data = right[
        [
            "entity_id",
            "name_norm",
            "address_norm",
            "country_norm"
        ]
    ].drop_duplicates("entity_id")

    result = candidates.merge(
        left_data,
        left_on="entity_id_left",
        right_on="entity_id",
        how="left"
    ).drop(columns=["entity_id"])

    result = result.merge(
        right_data,
        left_on="entity_id_right",
        right_on="entity_id",
        how="left",
        suffixes=("_left", "_right")
    ).drop(columns=["entity_id"])

    # Exact matches
    result["country_match"] = (
        result["country_norm_left"] == result["country_norm_right"]
    ).astype(int)

    result["name_exact"] = (
        result["name_norm_left"] == result["name_norm_right"]
    ).astype(int)

    result["address_exact"] = (
        result["address_norm_left"] == result["address_norm_right"]
    ).astype(int)

    # Prefix / suffix comparisons
    result["name_prefix_match"] = (
        result["name_norm_left"].str[:4] ==
        result["name_norm_right"].str[:4]
    ).astype(int)

    result["name_suffix_match"] = (
        result["name_norm_left"].str[-4:] ==
        result["name_norm_right"].str[-4:]
    ).astype(int)

    result["address_prefix_match"] = (
        result["address_norm_left"].str[:10] ==
        result["address_norm_right"].str[:10]
    ).astype(int)

    # Length ratios
    result["name_length_ratio"] = (
        result["name_norm_left"].str.len() /
        result["name_norm_right"].str.len().replace(0, 1)
    ).clip(upper=1)

    result["address_length_ratio"] = (
        result["address_norm_left"].str.len() /
        result["address_norm_right"].str.len().replace(0, 1)
    ).clip(upper=1)

    return result


# Separate candidates by source
candidates_s2 = train_candidates[
    train_candidates["source"] == "S2"
].copy()

candidates_s3 = train_candidates[
    train_candidates["source"] == "S3"
].copy()

print("Creating S1-S2 features...")
features_s2 = add_features(candidates_s2, train_s1, train_s2)

print("Creating S1-S3 features...")
features_s3 = add_features(candidates_s3, train_s1, train_s3)

# Combine
train_features = pd.concat(
    [features_s2, features_s3],
    ignore_index=True
)

print("Feature creation completed.")
print("Total feature rows:", len(train_features))
```

```python
# Prepare training data for the matching model

feature_columns = [
    "country_match",
    "name_exact",
    "address_exact",
    "name_prefix_match",
    "name_suffix_match",
    "address_prefix_match",
    "name_length_ratio",
    "address_length_ratio"
]

X = train_features[feature_columns].fillna(0)
y = train_features["label"].astype(int)

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
print("Positive labels:", (y == 1).sum())
print("Negative labels:", (y == 0).sum())
```

```python
# Train the matching model

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=500,
    class_weight="balanced",
    random_state=42
)

print("Training model...")

model.fit(X, y)

print("Model training completed successfully.")
```

```python
# Validate the model

from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score

# Use a smaller validation sample to keep computation manageable
X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.02,
    random_state=42,
    stratify=y
)

print("Validation data shape:", X_val.shape)

# Predict probabilities
val_prob = model.predict_proba(X_val)[:, 1]

print("Validation prediction completed.")
```

```python
# Find the best probability threshold using F0.5 score

import numpy as np
from sklearn.metrics import fbeta_score

thresholds = np.arange(0.10, 0.96, 0.05)

results = []

for threshold in thresholds:
    val_pred = (val_prob >= threshold).astype(int)

    score = fbeta_score(
        y_val,
        val_pred,
        beta=0.5,
        zero_division=0
    )

    results.append((threshold, score))

best_threshold, best_f05 = max(
    results,
    key=lambda x: x[1]
)

print("Best threshold:", round(best_threshold, 2))
print("Best F0.5 score:", round(best_f05, 4))
```

```python
# Generate candidate pairs for the test data

print("Generating test S1-S2 candidates...")

test_candidates_s1_s2 = make_candidates(test_s1, test_s2)

print("Generating test S1-S3 candidates...")

test_candidates_s1_s3 = make_candidates(test_s1, test_s3)

print("Test S1-S2 candidates:", len(test_candidates_s1_s2))
print("Test S1-S3 candidates:", len(test_candidates_s1_s3))
```

```python
# Create features for test candidates

test_candidates_s2 = test_candidates_s1_s2.copy()
test_candidates_s2["source"] = "S2"

test_candidates_s3 = test_candidates_s1_s3.copy()
test_candidates_s3["source"] = "S3"

print("Creating test S1-S2 features...")
test_features_s2 = add_features(
    test_candidates_s2,
    test_s1,
    test_s2
)

print("Creating test S1-S3 features...")
test_features_s3 = add_features(
    test_candidates_s3,
    test_s1,
    test_s3
)

test_features = pd.concat(
    [test_features_s2, test_features_s3],
    ignore_index=True
)

print("Test feature creation completed.")
print("Total test feature rows:", len(test_features))
```

```python
# Score test candidates using the trained model

test_X = test_features[feature_columns].fillna(0)

print("Scoring test candidates...")

test_features["match_probability"] = model.predict_proba(test_X)[:, 1]

# Keep candidates above the best validation threshold
final_matches = test_features[
    test_features["match_probability"] >= best_threshold
].copy()

print("Scoring completed.")
print("Total test candidates:", len(test_features))
print("Final predicted matches:", len(final_matches))
```

```python
# Create final output files

import os

OUTPUT_DIR = "/kaggle/working/output"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# --------------------------------------------------
# 1. matching_results.tsv
# --------------------------------------------------

# Group predicted matches by Source 1 entity
matching_results = (
    final_matches
    .groupby("entity_id_left")["entity_id_right"]
    .apply(lambda x: ",".join(x.astype(str)))
    .reset_index()
)

matching_results.columns = [
    "source1_entity_id",
    "matched_entity_ids"
]

# Make sure every test Source 1 entity is present
matching_results = test_s1[
    ["entity_id"]
].rename(
    columns={"entity_id": "source1_entity_id"}
).merge(
    matching_results,
    on="source1_entity_id",
    how="left"
)

matching_results["matched_entity_ids"] = (
    matching_results["matched_entity_ids"]
    .fillna("")
)

matching_results_path = f"{OUTPUT_DIR}/matching_results.tsv"

matching_results.to_csv(
    matching_results_path,
    sep="\t",
    index=False
)

# --------------------------------------------------
# 2. candidate_pairs.tsv
# --------------------------------------------------

candidate_pairs = test_features[
    ["entity_id_left", "entity_id_right", "source"]
].copy()

candidate_pairs.columns = [
    "source1_entity_id",
    "matched_entity_id",
    "source"
]

candidate_pairs_path = f"{OUTPUT_DIR}/candidate_pairs.tsv"

candidate_pairs.to_csv(
    candidate_pairs_path,
    sep="\t",
    index=False
)

print("Output files created successfully.")
print("Matching results:", matching_results_path)
print("Candidate pairs:", candidate_pairs_path)

print("\nMatching results rows:", len(matching_results))
print("Candidate pairs rows:", len(candidate_pairs))
```
